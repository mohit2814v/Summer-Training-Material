```python
import pandas as pd

students = pd.DataFrame({
    "Name": [
        "Alice", "Bob", "Charles", "David", "Emma",
        "Frank", "Grace", "Henry", "Ivy", "Jack"
    ],
    "Age": [18, 19, 20, 18, 21, 19, 20, 18, 21, 19],
    "Marks": [85, 92, 78, 88, 95, 81, 76, 89, 93, 84]
})

print(students)
```

```python
print(students.head())
```

```python
print(students[["Name", "Marks"]])
```

```python
students["> 70"] =students["Marks"] > 70
print(students)
```

```python
students.sort_values("Marks", inplace= True , ascending= False)
print(students)
```

```python
students["Result"] = students["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
print(students)
```

```python
print(students, end="\n")
print(students["Marks"].mean())
```

```python
print(students["Marks"].max())
```

```python
print((students["Result"] == "Pass").sum())
```

```python
students.set_index("Name", inplace=True)
print(students)
```

```python
print(students.columns)
```

```python
students.rename(columns={"Marks":"Score"} ,inplace =True)
print(students)
```
