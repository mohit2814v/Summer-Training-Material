```python
def sum_all(*args):
  total =0
  for num in args:
    total += num
  return total
print(sum_all(1,2,3,4))
```

```python
def fact(n):
  if(n==0 or n==1):
    return(1)
  else:
    return n*fact(n-1)

fact(5)
```
