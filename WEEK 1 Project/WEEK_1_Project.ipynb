```python
''' This project introduces simple gradebook system that can store
student names and their scores for various assignments. The goal is to
calculate basic statistics (mean, median, max, min) for each student
and each assignment.'''
```

```python
# TASK 1 - Data Representation using Pandas DataFrame
```

```python
# Student data stored in a dictionary
import pandas as pd
data= pd.DataFrame({
    "Name": ["Alice", "Bob", "Charles", "David", "Emma"],
    "Math": [85, 92, 78, 88, 95],
    "Science": [90, 85, 80, 86, 91],
    "English": [88, 79, 84, 90, 87],
    "Computer": [95, 89, 92, 85, 96],
    "History": [76, 82, 75, 89, 84]
})
students=pd.DataFrame(data)
print(students)
```

```python
# TASK 2 -DATA LOADING
# creating a dummy csv file for saving the data
pd.DataFrame(data).to_csv("student_data.csv", index=False)
# stores the data into CSV without indexing
```

```python
# Loading Data from "student_data.csv"
df=pd.read_csv("student_data.csv")
print(df)
```

```python
# TASK 3- Calculating the average , median , maximum and minimum scores for :
# a. Each student across all assignments
# b. Each assignment across all students.
```

```python
# ANSWER a
# Finding the average , median , maximum and minimum scores for alice
alice = df.query("Name == 'Alice'").drop(columns=["Name"])

subject = alice.idxmax(axis=1).iloc[0]
max = alice.max(axis=1).iloc[0]

subject2 = alice.idxmin(axis=1).iloc[0]
min = alice.min(axis=1).iloc[0]


mean = alice.mean(axis=1).iloc[0]

print("Subject:", subject)
print("Maximum:", max)

print("Marks:", subject2)
print("Minimum:", min)

print()
print("Mean:", mean)
```

```python
# ANSWER b
# FInding the average , median , maximum and minimum scores across all students
# MAX MARKS
print("Math")
print(df.loc[df["Math"].idxmax(), ["Name", "Math"]])

print("\nScience")
print(df.loc[df["Science"].idxmax(), ["Name", "Science"]])

print("\nEnglish")
print(df.loc[df["English"].idxmax(), ["Name", "English"]])

print("\nComputer")
print(df.loc[df["Computer"].idxmax(), ["Name", "Computer"]])

print("\nHistory")
print(df.loc[df["History"].idxmax(), ["Name", "History"]])
```

```python
# MIN MARKS
print("Math")
print(df.loc[df["Math"].idxmin(), ["Name", "Math"]])

print("\nScience")
print(df.loc[df["Science"].idxmin(), ["Name", "Science"]])

print("\nEnglish")
print(df.loc[df["English"].idxmin(), ["Name", "English"]])

print("\nComputer")
print(df.loc[df["Computer"].idxmin(), ["Name", "Computer"]])

print("\nHistory")
print(df.loc[df["History"].idxmin(), ["Name", "History"]])
```

```python
# MEAN MARKS PER SUBJECT
print("Math")
print(df["Math"].mean())

print("\nScience")
print(df["Science"].mean())

print("\nEnglish")
print(df["English"].mean())

print("\nComputer")
print(df["Computer"].mean())

print("\nHistory")
print(df["History"].mean())
```

```python
subjects = ["Math", "Science", "English", "Computer", "History"]

df["Average"] = df[subjects].mean(axis=1)

print(df)
```

```python
print(df[df["Average"] == df["Average"].max()])
```
