```python
import numpy as np
A=np.array([1,2,3])
B=np.array([[1,2,3],[4,5,6]])
def multiply_loops(A, B):

  C = np.zeros((A.shape[0], B.shape[1]))
  for i in range(A.shape[1]):

      for j in range(B.shape[0]):

        C[i, j] = A[i, j] * B[j, i]
  return C



#OR

def multiply_vector(A, B):

  return A @ B

print(multiply_loops(A,B))
```

```python
# Convert the list to a NumPy array
sales = np.array([
    [200, 250, 300, 280, 260], # Shop 1
    [180, 220, 270, 260, 240], # Shop 2
    [210, 260, 310, 290, 270]  # Shop 3
])

# Task 1: Find total sales of each shop (sum across each row)
total_sales_per_shop = np.sum(sales, axis=1)
print(f"Total sales per shop: {total_sales_per_shop}")
# Output: [1290 1170 1340]

# Task 2: Find the shop with highest total sales
# np.argmax returns the index of the highest value (0, 1, or 2)
# We add 1 to make it human-readable (Shop 1, 2, or 3)
best_shop_index = np.argmax(total_sales_per_shop)
print(f"Shop with highest sales: Shop {best_shop_index + 1} (Total: {total_sales_per_shop[best_shop_index]})")
# Output: Shop 3 (Total: 1340)

# Task 3: Find the average sales per day (mean down each column)
avg_sales_per_day = np.mean(sales, axis=0)
print(f"Average sales per day: {avg_sales_per_day}")
# Output: [196.66666667 243.33333333 293.33333333 276.66666667 256.66666667]
```
